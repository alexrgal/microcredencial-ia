## Operadores aritméticos

Los siguientes operadores trabajan con valores numéricos:

| Operador | Descripción | Ejemplo | Resultado |
|:---:|----|----|----|
| `+` | Suma | `3 + 2` | `5` |
| `-` | Resta | `7 - 4` | `3` |
| `*` | Multiplicación | `3 * 5` | `15` |
| `/` | División (flotante) | `7 / 2` | `3.5` |
| `//` | División entera | `7 // 2` | `3` |
| `%` | Módulo (resto) | `7 % 3` | `1` |
| `**` | Potencia | `2 ** 3` | `8` |

## Operadores de comparación (relacionales)

Devuelven valores booleanos (`True`, `False`) con el resultado de la comparación de dos expresiones:

| Operador | Significado   | Ejemplo  | Resultado |
| :------: | ------------- | -------- | --------- |
|   `==`   | Igual a       | `3 == 3` | `True`    |
|   `!=`   | Distinto de   | `3 != 4` | `True`    |
|   `<`    | Menor que     | `2 < 5`  | `True`    |
|   `<=`   | Menor o igual | `3 <= 3` | `True`    |
|   `>`    | Mayor que     | `5 > 2`  | `True`    |
|   `>=`   | Mayor o igual | `5 >= 6` | `False`   |

## Operadores lógicos: `and`, `or`, `not`

Trabajan con valores booleanos (`True` o `False`), combinándolos mediante las siguientes reglas lógicas:

| Operador | Significado          | Lógica clásica            |
| :------: | -------------------- | ------------------------- |
|  `and`   | Conjunción           | $\text{A}\land \text{B}$  |
|   `or`   | Disyunción           | $\text{A}\lor \text{B}$   |
|   `^`    | Disyunción exclusiva | $\text{A}\oplus \text{B}$ |
|  `not`   | Negación             | $\lnot \text{A}$          |

## Precedencia y asociatividad de Operadores

Orden (simplificado) de mayor a menor precedencia:

 1. **Paréntesis**: `( )`
 2. **Exponenciación**: `**` (asociatividad derecha)
 3. **Unarios**: `+x`, `-x`, `~x`
 4. **Multiplicativos**: `*`, `/`, `//`, `%`
 5. **Aditivos**: `+`, `-`
 6. **Desplazamientos**: `<<`, `>>`
 7. **Bit a bit**: `&`, `^`, `|`
 8. **Comparación**: `<`, `<=`, `>`, `>=`, `==`, `!=`, `is`, `in`
 9. **Lógicos**: `not`, `and`,  `or`
10. **Asignación**: `=`, `+=`, `-=`, ...

```python
(1 + 2 * (3 + 4/5) + 6) * 7
```

Una expresión más compleja, el precio de un bono:

$$\text{bond\_price} = \frac{\text{pmt}}{r}  \cdot \left[ 1-\frac{1}{(1+\text{r})^\text{n}} \right] + \frac{\text {face\_value}}{(1+\text{r})^\text{n}}$$

```python
face_value = 1000
pmt = 50
r   = 0.04
n   = 10

bond_price = pmt / r * ( 1 - 1 / (1 + r) ** n ) + face_value / (1 + r) ** n
bond_price
```
