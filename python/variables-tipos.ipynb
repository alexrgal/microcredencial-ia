## Variables

Para que un **nombre de variable** sea **válido** debe cumplir las siguientes reglas:
 - Debe contener solamente letras minúsculas (`a-z`), mayúsculas (`A-Z`), números (`0-9`) y guiones bajos (`_`).
 - No puede empezar por un número.
 - No puede coincidir con el nombre de una palabra reservada del lenguaje.
 - Python distingue entre mayúsculas y minúsculas.

## Conversión de tipos (*casting*)

Permite cambiar el tipo de un valor utilizando las siguientes funciones integradas:

| Conversión | Descripción                           | Ejemplo    | Resultado |
| :--------- | :------------------------------------ | :--------- | --------- |
| `int(x)`   | Convierte a entero (trunca decimales) | `int(3.9)` | `3`       |
| `float(x)` | Convierte a flotante                  | `float(5)` | `5.0`     |
| `str(x)`   | Convierte a cadena                    | `str(10)`  | `"10"`    |
| `bool(x)`  | Convierte a booleano (falsy / truthy) | `bool("")` | `False`   |
