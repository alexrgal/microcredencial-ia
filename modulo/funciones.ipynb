# Funciones

Una función es un bloque de código que realiza una tarea específica, recibe datos de entrada como parámetros, y devuelve un resultado.


Las funciones aportan:
- Reutilización
- Abstracción
- Modularidad
- Legibilidad

```python
def saludar(nombre):
    return f'Hola, {nombre}'

saludar('Alex')
```

```python
saludar('Alicia')
```

---
## Parámetros y argumentos

```python
def suma(a, b):
    return a + b

suma(3, 4) 
```

---
## Parámetros con valor por defecto

```python
def potencia(base, exponente=2):
    return base ** exponente

potencia(3) 
```

```python
potencia(3, 3)
```

---
## Argumentos nombrados (*keyword arguments*)

```python
potencia(exponente=3, base=2) 
```

```python
potencia(base=2, exponente=3) 
```

---
## `*args` y `**kwargs`

```python
def suma_todo(*args):
    return sum(args)

suma_todo(2, 3)
```

```python
suma_todo(1, 2, 3, 4)
```

```python
def imprimir_info(**kwargs):
    for clave, valor in kwargs.items():
        print(f'{clave}: {valor}')

imprimir_info(nombre='Alex', edad=45, ciudad='Madrid')
```

---
## `return` vs. `print`

```python
def f():
    print(5)    # muestra 5, pero devuelve None

resultado = f() * 2   # Error
```

```python
def g():
    return 5    # devuelve 5, que se puede usar

resultado = g() * 2

resultado
```

## Devolver múltiples valores

```python
def minmax(lista):
    return min(lista), max(lista)

minimo, maximo = minmax([3, 1, 4, 1, 5])
```

```python
minimo
```

```python
maximo
```

## Alcance de variables (*scope*)


| Ámbito        | Descripción                                            |
| :------------ | :----------------------------------------------------- |
| **L**ocal     | Dentro de la función actual                            |
| **E**nclosing | En funciones que envuelven a la actual (*closures*)    |
| **G**lobal    | En el módulo o script actual                           |
| **B**uilt-in  | Nombres propios de Python (`len`, `print`, `range`...) |


```python
x = 10          # variable global

def f():
    x = 5       # variable local, distinta de la global
    return x

f()
```

```python
x 
```

`global` para acceder a variable global.

```python
contador = 0

def incrementar():
    global contador
    contador += 1
```

```python
incrementar()

contador
```

## Mutabilidad y paso de argumentos

```python
def añadir(lista):
    lista.append(99)

mi_lista = [1, 2, 3]
añadir(mi_lista)
mi_lista
```

```python
def duplicar(n):
    n = n * 2   # solo modifica la variable local

x = 5
duplicar(x)
x           
```

## Funciones como objetos

```python
def cuadrado(x):
    return x ** 2

f = cuadrado    # f apunta a la misma función
f(4)
```

```python
def calcular(func, valores):
    return [func(v) for v in valores]

valores = [1, 2, 3, 4]
calcular(cuadrado, valores)
```

---
## Funciones lambda

```python
cubo = lambda x: x ** 3
cubo(4)
```

Es útil como argumento de otras funciones.

```python
valores = [1, 2, 3, 4]

calcular(lambda x: x*2, valores)
```
