## Condicionales

Siguen la siguiente sintaxis:

```python
if condicion:
    # bloque si la condición es verdadera
    # ...
elif otra_condicion:
    # (opcional) cuando la condición anterior no se cumple pero esta sí
    # ...
else:
    # (opcional) si ninguna de las anteriores condiciones se cumple
    # ...
```

### Ejemplo

````python
nota = 4

if nota >= 9:
    print('sobresaliente')
elif nota >= 7:
    print('notable')
elif nota >= 5:
	print('aprobado')
else:
    print('suspenso')
````

### Ejemplo con operadores lógicos

````python
temperatura = 22

if temperatura > 20 and temperatura < 25:
    print('Qué tiempo más agradable')
else:
    print('No estoy cómodo')
````

## Bucle `for`

Sigue esta sintaxis:

```python
for variable in iterable:
    # bloque de código
    # ...
else:
    # (opcional) si termina sin interrupción
    # ...
```

### Ejemplo

````python
frutas = ['manzana', 'pera', 'naranja']

for fruta in frutas:
    print(fruta)
````

## Uso de `enumerate()` 

Iteramos a la vez por índice (posición) y por valor:

```python
for indice, elemento in enumerate(iterable):
    # bloque de código
    #
```

### Ejemplo

````python
frutas = ['manzana', 'pera', 'naranja']

for i, fruta in enumerate(frutas):
    print(f'{i}: {fruta}')
````

## Uso de `range()`
Sintaxis de `range()`:

| **Sintaxis**               | Valores que recorre                                   |
| :------------------------- | :---------------------------------------------------- |
| `range(fin)`               | desde `0` hasta `fin-1`.                              |
| `range(inicio, fin)`       | desde `inicio` hasta `fin-1`.                         |
| `range(inicio, fin, paso)` | desde `inicio` hasta `fin-1`en incrementos de `paso`. |

### Ejemplos

````python
for i in range(5):
    print(i)
````

````python
suma = 0
for impar in range(1, 10, 2):  # inicio, fin, paso
    print(impar)
    suma = suma + impar
    
print(f'la suma es {suma}')
````

## Bucle `while`

La sintaxis es:

```python
while condicion:
    # bloque de código
    # ...
else:
    # Opcional: si termina sin interrupción
    # ...
```

### Ejemplos


````python
contador = 1

while contador < 5:
	print(contador)
	contador += 1
````

### Ejemplo

Este programa realiza ingresos en una cuenta de ahorro y avisa cuando se alcanza un determinado objetivo de ahorro.

````python
saldo_cuenta    = 0
objetivo_ahorro = 1000

while saldo_cuenta < objetivo_ahorro:

	# pedimos cantidad a ingresar por teclado
    ingreso = float(input('¿Qué cantidad deseas depositar en la cuenta?'))

	# actualizamos el saldo    
    saldo_cuenta += ingreso
    print(f'Tienes {saldo_cuenta} euros en tu cuenta')

# el bucle while ha terminado con éxito
print(f'!Has alcanzado tu objetivo de ahorrar {objetivo_ahorro} euros!')
````

## Bucles anidados

Los bloques de código asociados a estructuras condicionales y bucles pueden, a su vez, **contener otras estructuras de control en su interior**. A este tipo de construcciones se las denomina **estructuras o bucles anidados**.

### Ejemplo

````python
for a in range(1, 4):
    for b in range(1, 4):
        producto = a * b
        print(f'{a} x {b} = {producto}')
    print('---')
````

## Control de bucles

### `break`

Interrumpe un bucle de forma inmediata.

#### Ejemplo

````python
for numero in range(10):
    if numero == 5:
	    # al llegar a 5, sale del bucle
        break
    print(numero)
````

### `continue`

Salta directamente a la siguiente iteración.

#### Ejemplo

````python
for numero in range(5):
    if numero == 2:
	    # al llegar a 2 pasa a la siguiente iteración
        continue
    print(numero)
````

## La instrucción de relleno `pass`

La instrucción `pass` se utiliza como **marcador de posición** cuando la sintaxis de Python exige un bloque de código, pero no se desea ejecutar ninguna acción.

No produce ningún efecto en tiempo de ejecución, pero **evita errores de sintaxis** y resulta especialmente útil durante el desarrollo para dejar "en blanco" bloques de código que se escribirán en otro momento.

### Ejemplo

En este ejemplo, queda claro el funcionamiento del programa aunque no hayamos escrito todavía el bloque de código para tratar las incidencias de severidad `Media` (hemos colocado un `pass` provisionalmente en su lugar). Todo lo demás, el caso de `Alto` y la cláusula `else`, funcionan correctamente.


````python
severidad = 'Alta'

if severidad == 'Alta':
	print('La incidencia es urgente')
elif severidad == 'Media':
	# TO DO: programaré esto en otro momento
    pass
else:
	print('La incidencia puede esperar')
````

## Expresiones condicionales (`if` en una línea)

Las **expresiones condicionales**, también conocidas como **expresiones ternarias**, permiten evaluar una condición y asignar un valor u otro **en una sola línea de código**.

```python
valor_si_verdadero if condicion else valor_si_falso
```

### Ejemplo

````python
edad = 20

adjetivo = 'adulto' if edad >= 18 else 'menor'

print(f'si tienes {edad} años eres {adjetivo}')
````
