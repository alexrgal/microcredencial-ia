## Ejercicio: CRUD de una librería

Escriba una aplicación que gestione el inventario de libros de una librería. Un libro viene identificado por su **título,** y nos gustaría guardar información adicional como su **autor**, el **año** de publicación, el **género** y el **precio**. Codifique la aplicación completa, incluyendo un menú, que satisfaga las siguientes opciones.

El programa debe procesar sucesivas peticiones del usuario hasta que este indique en el menú que desea salir.

1. **Añade un nuevo libro.**

```python
def añadir_libro (stock, ...)
```

Solicita al usuario los datos necesarios del nuevo libro y añádelo a `stock`, que es un diccionario con todos los libros actualmente en la tienda.

2. **Muestra por pantalla el _título_ y _precio_ de todos los libros de un _género_ elegido por el usuario.**

```python
def mostrar_libros_genero (stock, ...)
```

3. **Actualiza el _precio_ de un libro.**

```python
def actualizar_precio (stock, ...)
```

Solicita el título y el nuevo precio al usuario. Si el nuevo precio es inferior al anterior, muestra un mensaje de advertencia (y realiza el cambio de precio igualmente).

4. **Borra un libro.**

```python
def borrar_libro (stock, ...)
```

Solicita al usuario el título y borra el libro indicado del `stock`. Si no figura en el inventario, muestra un mensaje de error.

5. **Muestra por pantalla la suma de los _precios_ de los libros anteriores a un _año_ proporcionado por el usuario.**

```python
def sumar_anteriores (stock, ...)
```

6. **Imprime un menú con las opciones anteriores, incluyendo la posibilidad de abandonar el programa. Dependiendo de la elección del usuario, realiza la acción deseada.**

**Pista:** `stock` puede modelarse como un diccionario de diccionarios o un diccionario de listas.

````python
def añadir_libro(stock):

    titulo = input('Título del libro: ')
    autor = input('Autor: ')
    año = int(input('Año de publicación: '))
    genero = input('Género: ')
    precio = float(input('Precio: '))

    datos = { 'autor' : autor, 'año' : año, 'genero' : genero, 'precio' : precio }
    stock[titulo] = datos


def mostrar_libros_genero (stock):

    genero_buscado = input('Introduce el genero a buscar: ')

    for titulo, libro in stock.items():

        if libro['genero'] == genero_buscado:
            
            precio = libro['precio']
            print(f'{titulo}: {precio:.2f}')


def actualizar_precio(stock):

    titulo = input('Título del libro cuyo precio desea cambiar: ')
    nuevo_precio = float(input('Nuevo precio: '))

    libro = stock[titulo]

    if nuevo_precio < libro['precio']:
        print('ADVERTENCIA: el nuevo precio es menor')

    libro['precio'] = nuevo_precio


def borrar_libro(stock):

    titulo = input('Título del libro a borrar: ')

    if titulo in stock:
        stock.pop(titulo)
    else:
        print(f'El libro {titulo} no existe')


def sumar_anteriores(stock):

    año_hasta = int(input('Introduce el año hasta el cual sumaré precios: '))

    total = 0.0

    for libro in stock.values():

        if libro['año'] <= año_hasta:
            total += libro['precio']

    print(f'La suma de los precios de los libros publicados antes de {año_hasta} es {total:.2f}')
    

# stock = {}

stock = {  
    'Dune': {
	    'autor': 'Frank Herbert',
	    'año': 1965,
	    'genero': 'sci-fi',
	    'precio': 29.95},  
    'Fundación': {
	    'autor': 'Isaac Asimov',
	    'año': 1951,
	    'genero': 'sci-fi',
	    'precio': 34.95},  
    'La guía del autoestopista galáctico': {
	    'autor': 'Douglas Adams',
	    'año': 1979,
	    'genero': 'sci-fi',
	    'precio': 22.95}
}  

continuar = True

while continuar:
    print('Menú de gestión de inventario')
    print(' 1. Añadir libro nuevo')
    print(' 2. Mostrar libros por género')
    print(' 3. Actualizar precio de un libro')
    print(' 4. Borrar un libro')
    print(' 5. Sumar precios de libros anteriores a fecha')
    print(' 6. Salir')
    
    opcion = input('Seleccione una opción [1-6]: ')
    
    if opcion == '1':
        añadir_libro(stock)
    elif opcion == '2':
        mostrar_libros_genero(stock)
    elif opcion == '3':
        actualizar_precio(stock)
    elif opcion == '4':
        borrar_libro(stock)
    elif opcion == '5':
        sumar_anteriores(stock)
    elif opcion == '6':
        continuar = False
    else:
        print('Opción no válida. Por favor elige un número entre 1 y 6.')
    
````
