## Ejercicio: Radar de tramo

Trabajas en la Dirección General de Tráfico (DGT) y te han encargado un programa para controlar la velocidad de los vehículos en un tramo de carretera.

El programa debe realizar lo siguiente:

1. Solicitar al usuario la **distancia en kilómetros** del tramo.
2. Solicitar el **tiempo en segundos** que un coche ha tardado en recorrerlo.
3. A partir de los datos anteriores, calcular la **velocidad media en kilómetros por hora**.
4. **Mostrar un mensaje** indicando si el conductor ha respetado el límite de velocidad de 120 km/h o si debe ser multado.

**Pista:**

$$\text{Velocidad media [km/h]} = \frac{\text{Distancia [km]}}{\text{Tiempo [h]}}$$

$$\text{Tiempo [h]} = \frac{\text{Tiempo [s]}}{3600}$$

*(8 líneas de código)*

```python
distancia = float(input('Introduzca la distancia'))
tiempo_s  = float(input('Introduzca el tiempo'))

tiempo_h  = tiempo_s / 3600

velocidad = distancia / tiempo_h

print(f'TRAZA: {distancia} {tiempo_s} {tiempo_h} {velocidad}')

if velocidad <= 120:
    print(f'No multar, la velocidad es {velocidad} Km/h')
else:
    print(f'Multa, la velocidad es {velocidad} Km/h')
```

## Ejercicio: Impuestos

Escribe una función que reciba como entrada los ingresos de una persona y calcule los impuestos que debe pagar según esta tabla:

| Ingresos | Impuestos % |
|----|----|
| Primeros 5.000 euros | 0% |
| Siguientes 30.000 euros | 10% |
| Siguientes 65.000 euros | 20% |
| Resto | 33% |

*(9 líneas de código)*

```python
def impuestos(ingresos):

    if ingresos <= 5000:
        return 0

    elif ingresos <= 35000:
        return 0 + 0.1 * (ingresos - 5000)

    elif ingresos <= 100000:
        return 0 + 3000 + 0.2 * (ingresos - 5000 - 30000)

    else:
        return 0 + 3000 + 13000 + 0.33 * (ingresos - 5000 - 30000 - 65000)

# probamos
impuestos(110000)
```
