# Excepciones

Las **excepciones** son errores que pueden ocurrir durante la ejecución de un programa.

Se capturan y gestionan con bloques `try` / `except`.

---
## Errores de sintaxis

```python
print('hola']
```

---
## Error en tiempo de ejecución (Excepción)

```python
print(10 / 0)   
```

---
## Estructura básica: `try`/ `except`

```python
try:
    # Código que puede causar un error
    x = 10 / 0

except ZeroDivisionError:
    print('No se puede dividir entre cero.')

print('El programa continúa...')
```

---
## Excepciones más comunes


| Excepción           | Causa habitual                                           |
| ------------------- | -------------------------------------------------------- |
| `ZeroDivisionError` | División entre cero                                      |
| `ValueError`        | Valor incorrecto para la operación (`int('abc')`)        |
| `TypeError`         | Tipo incorrecto (`'hola' + 5`)                           |
| `IndexError`        | Índice fuera de rango en una lista                       |
| `KeyError`          | Clave inexistente en un diccionario                      |
| `FileNotFoundError` | Archivo que no existe                                    |
| `AttributeError`    | Atributo o método inexistente en un objeto               |
| `NameError`         | Variable no definida                                     |
| `ImportError`       | Módulo que no se puede importar                          |
| `RecursionError`    | Recursión demasiado profunda (función recursiva sin fin) |
| ...                 | ... |

## Capturar múltiples tipos de excepción

```python
try:
    numero = int(input('Introduce un número: '))
    resultado = 10 / numero

except ValueError:
    print('Debes introducir un número válido.')
except ZeroDivisionError:
    print('No se puede dividir entre cero.')
```

---
## Capturar cualquier excepción

```python
try:
    x = int('abc')
    
except Exception as e:
    print(f'Mensaje de error: {e}')
```

---
## Uso de `else`

```python
try:
    x = int(input('Introduce un número entero: '))
    
except ValueError:
    print('No es un número entero válido.')
    
else:
    print(f'Número {x} introducido correctamente')
```

---
## Uso de `finally`

```python
def dividir(a, b):

    try:
        resultado = a / b
        
    except ZeroDivisionError:
        print('Error: división por cero.')
        
    else:
        print(f'Resultado: {resultado}')
        
    finally:
        print('Operación finalizada.')


dividir(10, 2)
```

```python
dividir(10, 0)
```

---
## Agrupar múltiples excepciones

```python
try:
    x = int(input('Número: '))
    print(10 / x)
    
except (ValueError, ZeroDivisionError):
    print('Error: valor no válido o división por cero.')
```

---
## Lanzar excepciones manualmente (`raise`)

```python
def comprobar_edad(edad):
    if edad < 0:
        raise ValueError('La edad no puede ser negativa.')
    return edad

try:
    comprobar_edad(-5)
    
except ValueError as e:
    print(f'Error: {e}')
```

---
## Re-lanzar una excepción

```python
try:
    x = int('abc')
    
except ValueError as e:
    print('Registrando el error...')
    raise  # Vuelve a lanzar la misma excepción
```

---
## Encadenar excepciones (`raise` / `from`)

```python
try:
    datos = int('abc')
    
except ValueError as e:
    raise RuntimeError('Error al procesar los datos.') from e
```

## Buenas prácticas

 - Sé específico
 - No silencies errores
 - No abuses de las excepciones
 - Proporciona mensajes claros
