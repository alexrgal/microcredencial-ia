## F-strings (cadenas formateadas)

````python
nombre   = 'Ana'
edad     = 30
estatura = 1.7525

"Hola " + nombre + ", tu edad es " + str(edad) + " y tu estatura " + str(round(estatura, 2))
````

````python
f"Hola {nombre}, tu edad es {edad} y tu estatura {estatura:.2f}"
````

## Métodos comunes para trabajar con cadenas

Python ofrece un amplio abanico de funciones para trabajar con cadenas. Estos serían algunos ejemplos:

| Método             | Descripción                                | Ejemplo                     | Resultado |
| ------------------ | ------------------------------------------ | --------------------------- | --------- |
| `upper()`          | Convierte a mayúsculas                     | `'hola'.upper()`            | `HOLA`    |
| `lower()`          | Convierte a minúsculas                     | `'HOLA'.lower()`            | `hola`    |
| `find(sub)`        | Devuelve el índice de la primera aparición | `'banana'.find('a')`        | `1`       |
| `rfind(sub)`       | Índice de la última aparición              | `'banana'.rfind('a')`       | `5`       |
| `count(sub)`       | Cuenta apariciones                         | `'banana'.count('a')`       | `3`       |
| `len(cad)`         | Longitud de la cadena                      | `len('Python')`             | `6`       |
| `replace(a, b)`    | Sustituye texto                            | `'hola'.replace('a', 'o')`  | `holo`    |
| `strip()`          | Elimina espacios en extremos               | `'  hola  '"'.strip()`      | `hola`    |
| `startswith(pref)` | Comprueba prefijo                          | `'python'.startswith('py')` | `True`    |
| `endswith(suf)`    | Comprueba sufijo                           | `'script'.endswith('pt')`   | `True`    |


````python
titulo = 'Título de mi trabajo'

titulo.upper()
````

````python
# Nota: python empieza a contar en 0
ALFABETO = 'abcdefghijklmnopqrstuvwxyz'

ALFABETO.find('m')
````

````python
profesion = 'marinero'
    
profesion.replace('ar', 'ol')
````

### Entrada salida básica

| **Método**                 | **Descripción**                                                                                           |
| :------------------------- | :-------------------------------------------------------------------------------------------------------- |
| `entrada = input(mensaje)` | Muestra `mensaje` por pantalla y guarda lo que el usuario introduce por teclado en la variable `entrada'. |
| `print(salida)`            | Muestra `salida` por pantalla.                                                                            |

Un pequeño programa que:
1. Pide una palabra.
2. El usuario la introduce por teclado.
3. Cuenta el número de caracteres en la palabra.
4. Imprime el resultado.

````python
entrada = input('introduce una palabra:')

entrada = entrada.strip()   # eliminamos espacios innecesarios
longitud = len(entrada)     # calculamos la longitud

print(f'La palabra "{entrada}" tiene {longitud} caracteres')
````
