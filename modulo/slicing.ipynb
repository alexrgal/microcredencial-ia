# Slicing

El **slicing** en Python permite extraer subconjuntos de una secuencia ordenada (como listas, tuplas, cadenas u otros formatos tabulares) indicando un rango mediante la sintaxis:

`secuencia [inicio : fin : paso]`

Donde:
- `inicio` es el índice desde el que se empieza (inclusive).
- `fin` es el índice donde se termina (exclusive).
- `paso` indica el salto entre elementos.

La sintaxis es flexible y, si se omite alguno de estos valores, Python asume por defecto que `inicio` es `0`, `fin` es el final de la secuencia, y `paso` es `1`.

```python
letras = ['A', 'B', 'C', 'D', 'E', 'F']
# índice:  0    1    2    3    4    5
```

Desde la posición 2, inclusive, hasta la 5, exclusive.

```python
letras[2:5]     # ['C', 'D', 'E']
```

Desde el inicio, hasta la posición 2, exclusive.

```python
letras[ :2]     # ['A', 'B']
```

Desde la posición 3, inclusive, hasta el final.

```python
letras[3: ]     # ['D', 'E', 'F']
```

---
## Índices negativos

Los índices negativos permiten contar desde el final de la secuencia, es decir, de derecha a izquierda. El último elemento tiene índice `-1`, el penúltimo `-2`, y así sucesivamente.

Se pueden combinar índices negativos y positivos dentro de la misma operación.

```python
letras = ['A', 'B', 'C', 'D', 'E', 'F']
# índice: -6   -5   -4   -3   -2   -1
```

Desde la posición -4, inclusive, hasta la -1, exclusive.

```python
letras[-4:-1]   # ['C', 'D', 'E']
```

Desde el inicio, hasta la posición -4, exclusive.

```python
letras[ :-4]     # ['A', 'B']
```

Desde la posición -3, inclusive, hasta el final.

```python
letras[-3: ]     # ['D', 'E', 'F']
```

---
## Uso de `paso` (step)

El parámetro `paso` define cada cuántos elementos se selecciona uno. Puede ser **negativo**, lo que Python interpreta como recorrer la secuencia del revés (de derecha a izquierda)

```python
letras = ['A', 'B', 'C', 'D', 'E', 'F']
```

Desde el inicio hasta el final, seleccionar un elemento de cada dos

```python
letras[ : :2]   # ['A', 'C', 'E']
```

Desde la posición 1, inclusive, hasta el final, seleccionando de dos en dos

```python
letras[1: :2]   # ['B', 'D', 'F']
```

Desde el final hasta el inicio, recorrer de uno en uno hacia atrás

```python
letras[ : :-1]   # ['F', 'E', 'D', 'C', 'B', 'A']
```

Desde la posición 5, inclusive, hasta la posición 1, exclusive, recorrer de uno en uno hacia atrás

```python
letras[5:1:-1]   # ['F', 'E', 'D', 'C']
```

---
## Slicing con cadenas

El **slicing** funciona del mismo modo en cadenas (`strings`) ya que son, en esencia, secuencias de caracteres.

```python
palabra = "empresa"
```

Desde la posición 1, inclusive, hasta la 4, exclusive

```python
palabra[1:4]     # 'mpr'
```

Recorrer la cadena en sentido inverso

```python
palabra[ : :-1]  # 'asrepme'
```

Seleccionar un caracter de cada 3.

```python
palabra[ : :3]  # 'era'
```

## Combinación con `len()`

A veces puede ser interesante emplear `len()` para hacer cortes dinámicos que se ajustan a la longitud de la secuencia:

```python
datos = [10, 20, 30, 40, 50, 60]

# usamos división entera para que la posición de corte sea entera
mitad = len(datos) // 2

datos[:mitad]   # Primera mitad
```

```python
datos[mitad:]   # Segunda mitad
```
