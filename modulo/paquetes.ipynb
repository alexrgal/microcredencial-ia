# Paquetes

 - Un **módulo** es un archivo `.py` con funciones, clases y variables reutilizables.
 - Un **paquete** es una colección de módulos relacionados.
 - La instrucción `import` permite incorporar módulos a tu programa.
 - Con `from ... import ...` se importan solo los elementos concretos que necesitas.
 - Con `import ... as ...` se asigna un alias, algo muy habitual en el ecosistema de Data Science.
 - Python incluye una extensa **biblioteca estándar** lista para usar.
 - Los paquetes de terceros se instalan con `pip`.

---
## Uso de `import ...`

```python
import math

math.sqrt(16)
```

```python
math.pi
```

---
## Uso de `from ... import ...`

```python
from math import sqrt, pi

sqrt(16)
```

```python
pi
```

## La biblioteca estándar

Python incluye una extensa **biblioteca estándar**. Algunos ejemplos:

| Módulo       | Para qué sirve                                              |
| :----------- | :---------------------------------------------------------- |
| `math`       | Funciones matemáticas: `sqrt`, `log`, `sin`, `pi`...        |
| `random`     | Generación de números aleatorios                            |
| `statistics` | Media, mediana, desviación típica...                        |
| `datetime`   | Fechas, horas e intervalos de tiempo                        |
| `os`         | Interacción con el sistema operativo (rutas, archivos...)   |
| `pathlib`    | Manipulación de rutas de archivos de forma orientada a objetos |
| `json`       | Lectura y escritura de datos en formato JSON                |
| `csv`        | Lectura y escritura de archivos CSV                         |
| `re`         | Expresiones regulares para buscar y transformar texto       |
| `collections`| Estructuras de datos adicionales: `Counter`, `defaultdict`... |
| `itertools`  | Herramientas para trabajar con iteradores de forma eficiente |
| ...          | **... ¡Y mucho más!** |

```python
from datetime import date

fecha = date.today()
print(f'hoy es {fecha}')
```
