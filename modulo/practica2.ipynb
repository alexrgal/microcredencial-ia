## Ejercicio: Interés compuesto
Un profesor de programación tiene dos hijos pequeños. Para fomentar el ahorro, se ofrece a pagarles un interés mensual del 1% sobre lo que tengan ahorrado en su hucha. Estos intereses se introducen de nuevo en la hucha, incrementando así el capital mes a mes (interés compuesto).  

$$\text{capital}_{\;t+1} = \text{capital}_{\;t} \times (1+\text{interes})$$

Escribe un programa que pida por teclado el capital inicial y a continuación imprima su evolución durante los siguientes 6 meses.

````python
INTERES = 0.01
MESES = 6

capital = float(input('Introduzca capital inicial: '))

for mes in range(MESES):

    capital = capital * (1 + INTERES)

    print(f'{mes+1} {capital:.2f}')
````

## Ejercicio: Retornos bursátiles

El retorno de una acción se puede calcular como el porcentaje de variación de su precio entre dos momentos de tiempo.

$$\text{Retorno} = \frac{\text{Precio Final} - \text{Precio Inicial}}{\text{Precio Inicial}} \times 100\%$$

Por ejemplo, si el precio de una acción era 100 euros un año y 120 euros al siguiente, el retorno anual sería del 20%.

$$\text{Retorno} = \frac{120 - 100}{100} \times 100\%= 20\%$$

Dada esta serie de precios anuales (empezando en el año 0):

```python
PRECIOS = [80, 100, 120, 90, 135, 108]
```

Escribe un programa que:
1. Calcule el retorno anual de la acción para cada año, usando el precio del año anterior como precio inicial.
2. Imprima por pantalla el año y el retorno que se acaba de calcular.

La salida esperada es la siguiente:
```
Año 1: Retorno = 25.00%
Año 2: Retorno = 20.00%
Año 3: Retorno = -25.00%
...
```

*(3 líneas de código, sin contar la lista de precios)*

````python
# Dato
PRECIOS = [80, 100, 120, 90, 135, 108]

for i in range(1, len(PRECIOS)):

    retorno = (PRECIOS[i] - PRECIOS[i-1]) / PRECIOS[i-1] * 100
    print(f'Año {i}: Retorno = {retorno:.2f}%')
````

Un supermercado tiene los siguientes precios por unidad de sus productos:  
  
```python  
PRECIOS = {

    'Huevos': 2.20,
```  

Escriba una función `calcular_precio(precios, carrito)` que:  
  
1. Calcule el total a pagar por un cliente dado un `carrito` de la compra, representado como una lista de productos. Por ejemplo, si el `carrito` es `['Pan', 'Pan', 'Leche', 'Leche', 'Huevos']`, el total a pagar sería:  
$$0.50 + 0.50 + 1.90 + 1.90 + 2.20 = 7.00 \; \text{euros}$$  
2. Si el carrito contiene un producto que no está en el diccionario de precios, la función debe mostrar un mensaje de error indicando que el producto es desconocido y continuar con el cálculo del total ignorando ese producto.

````python
# Dato  
PRECIOS = {  
    'Pan': 0.50,  
    'Leche': 1.90,  
    'Huevos': 2.20,  
    'Queso': 3.50,  
    'Jamón': 5.00  
} 

def calcular_precio(precios, carrito):

    total = 0.0
    
    for producto in carrito:

        if producto in precios:
            total = total + precios[producto]
        else:
            print(f'El producto {producto} no existe en el catálogo. Se ignora.')

    return total

# Prueba unitaria
carrito_prueba = ['Pan', 'Pan', 'Leche', 'Plátano']
calcular_precio(PRECIOS, carrito_prueba)
        


````

### Ejercicio: Selección de proyectos

El *payback* (o plazo de recuperación) es el tiempo que un proyecto tarda en generar suficientes ingresos para recuperar la inversión inicial. Es decir, si el *payback* es 3 años, significa que los flujos de caja recibidos en los años 1, 2, y 3 son iguales (o superiores) a la inversión inicial del año 0.

Escrito de otra forma, que quizás os resulte más útil al programar:

$$
\sum_{t=0}^\text{payback} \text{cashflow}_t \ge 0
$$

Escribe una función `payback(cashflows)` que calcule el *payback* de una lista de flujos de caja anuales. La inversión inicial ocupará la primera posición en la lista y tendrá signo negativo.

El Valor Actual Neto (VAN), denominado en inglés *Net Present Value* (NPV), es la suma en el presente de los flujos de caja futuros de un proyecto, descontados según una tasa de descuento $r$. Todo proyecto financiero debe tener un $NPV \ge 0$ para ser viable y aceptado.

$$
NPV = \sum_{t=0}^N \frac{\text{cashflow}_t}{(1+r)^t}
$$

Escribe una función `npv(cashflows, r)` que calcule el NPV de una lista de flujos de caja anuales. La inversión inicial ocupará la primera posición en la lista y tendrá signo negativo.

Empleando las funciones `payback()` y `npv()` de los ejercicios anteriores, escribe un programa que recorra una lista de proyectos y muestre en forma tabular su *payback* y NPV.

```python
R = 0.10

PROYECTOS = {
  'restaurante' : [-1000, 250, 250,  300,  300,  300],
  'joyería'     : [-3500, 500, 700, 1000, 1200, 1500],
  'golosinas'   : [ -100,  25,  25,   25,   25,   25]
}
```



````python
def payback(cashflows):

    acumulado = 0

    for t, cf in enumerate(cashflows):
        acumulado += cf
        if acumulado >= 0:
            return t

    return 'error'

def npv(cashflows, r):

    acumulado = 0

    for t, cf in enumerate(cashflows):
        acumulado += cf / (1+r)**t

    return acumulado


R = 0.10

PROYECTOS = {
  'restaurante' : [-1000, 250, 250,  300,  300,  300],
  'joyería'     : [-3500, 500, 700, 1000, 1200, 1500],
  'golosinas'   : [ -100,  25,  25,   25,   25,   25]
}

for nombre, cashflows in PROYECTOS.items():

    res_payback = payback(cashflows)
    res_npv = npv(cashflows, R)

    print(f'{nombre:16} {res_payback:2}\t{res_npv:.2f}')
````
