## Listas
Una **lista** es una estructura de datos que permite almacenar una colección ordenada de elementos, los cuales pueden ser de diferentes tipos.

Las listas son **mutables**, lo que significa que se pueden modificar después de su creación (e.g., añadiendo, cambiando o eliminando elementos).

Se definen entre corchetes `[ ... ]` y los elementos se separan con comas (`,`).

```python
frutas = ['manzana', 'pera', 'plátano']
numeros = [1, 2, 3, 4, 5]
mixta = [1, 'hola', True, 3.14]
vacia = [ ]
```

```python
mixta
```

---
## Acceso a elementos

Los elementos de una lista se acceden mediante su índice, comenzando desde `0` para el primer elemento.

Si empleamos índices negativos, Python empezará a contar desde el final de la lista hacia el principio. El índice `-1` se refiere al último elemento.

```python
frutas[0]
```

```python
frutas[1]
```

```python
frutas[-1]
```

---
## Modificación de elemento

Se puede modificar un elemento de la lista asignando un nuevo valor en su posición.

```python
frutas[1] = 'naranja'
```

```python
frutas
```

---
## Longitud

El número de elementos que contiene una lista se obtiene con la función `len()`.

```python
len(frutas)
```

---
## Comprobación de contenido

Se puede verificar si un elemento está presente en una lista utilizando el operador de pertenencia `in`.

```python
"pera" in frutas      # False
```

```python
"naranja" in frutas   # True
```

---
## Métodos útiles de las listas

A continuación se enumeran algunos métodos muy útiles para trabajar con listas:

| Método         | Descripción                                                  |
| :------------- | :----------------------------------------------------------- |
| `append(x)`    | Añade un elemento al final.                                  |
| `count(x)`     | Cuenta apariciones de `x`.                                   |
| `index(x)`     | Devuelve la posición de `x`.                                 |
| `insert(i, x)` | Inserta `x` en la posición `i`, desplazando los demás.       |
| `remove(x)`    | Elimina la primera ocurrencia de `x`.                        |
| `reverse()`    | Invierte el orden.                                           |
| `pop(i)`       | Elimina y devuelve el elemento `i` (último si no se indica). |
| `sort()`       | Ordena la lista.                                             |


```python
numeros = [3, 9, 3, 4]
```

```python
numeros.append(5)    # [3, 9, 3, 4, 5]

numeros
```

```python
numeros.count(3)     # 2 apariciones
```

```python
numeros.index(4)     # 3
```

```python
numeros.insert(1, 6) # [3, 6, 9, 3, 4, 5]

numeros
```

```python
numeros.remove(4)    # [3, 6, 9, 3, 5]

numeros
```

```python
numeros.reverse()    # [5, 3, 9, 6, 3]

numeros
```

```python
numeros.pop()        # devuelve 3, la lista quedaría [5, 3, 9, 6]

numeros
```

```python
numeros.sort()       # [3, 5, 6, 9]

numeros
```

## Recorrido de listas

Una lista es **iterable**, lo que significa que se puede recorrer elemento a elemento utilizando un bucle `for`.

```python
frutas
```

```python
for fruta in frutas:
    print(fruta)
```

De forma alternativa, también se puede iterar con índices:

```python
longitud = len(frutas)

longitud
```

```python
for i in range(longitud):
    print(i, frutas[i])
```

Y también con `enumerate()` que proporciona simultáneamente la posición y el valor de los elementos de una lista. El siguiente código produce la misma salida que el ejemplo anterior:

```python
for i, elemento in enumerate(frutas):
	print(i, elemento)
```

## Copias de listas

Al asignar una lista a otra variable, ambas variables apuntan a la misma lista en memoria. Por lo tanto, modificar una de ellas afectará a la otra.

```python
lista_a = [1, 2, 3]

lista_b = lista_a     # copia por referencia

lista_b.append(4)
```

```python
lista_a               # [1, 2, 3, 4]
```

Para crear una copia independiente de una lista, podemos usar `copy()`. En este caso, modificar `lista_c`, que es un copia, no cambia la lista original en `lista_a`.

```python
lista_c = lista_a.copy()  # copia independiente

lista_c.pop()
```

```python
lista_a  # [1, 2, 3, 4]
```

```python
lista_c  # [1, 2, 3]
```

## Comprensiones de listas

Las **comprensiones de listas** son una forma concisa y eficiente de crear nuevas listas a partir de iterables existentes, aplicando una expresión a cada elemento y opcionalmente filtrando elementos con una condición.


```python
cuadrados = [x**2 for x in range(6)]

cuadrados       # [0, 1, 4, 9, 16, 25]
```

Un ejemplo de comprensión que incluye una condición:

```python
pares = [x for x in range(10) if x % 2 == 0]

pares           # [0, 2, 4, 6, 8]
```

## Tuplas

Una **tupla** es una estructura de datos similar a una lista, pero es **inmutable**, lo que significa que no se pueden modificar después de su creación.

Las tuplas se definen entre paréntesis `( ... )` y los elementos se separan con comas (`,`).

```python
coordenadas = (10, 20)
colores = ('rojo', 'verde', 'azul')
```

```python
coordenadas
```

---
## Acceso a elementos

Al igual que las listas, los elementos de una tupla se acceden mediante su índice, comenzando desde `0` para el primer elemento.

```python
coordenadas[0]
```

---
## Inmutabilidad

Cualquier intento de modificar un elemento de una tupla provocará un error, ya que las tuplas no permiten cambios después de su creación.

```python
coordenadas[0] = 5  # ERROR: las tuplas no pueden modificarse
```

---
## Conversión entre listas y tuplas

Se pueden convertir listas en tuplas y viceversa utilizando las funciones `tuple()` y `list()`, respectivamente.

```python
frutas = ['pera', 'plátano']
tupla_frutas = tuple(frutas)

tupla_frutas        # ('pera', 'plátano')
```

```python
lista_frutas = list(tupla_frutas)

lista_frutas        # ['pera', 'plátano']
```

## Desempaquetado de tuplas y listas

El **desempaquetado** es una técnica que permite asignar los elementos de una tupla o lista a variables individuales de forma sencilla.

```python
coordenadas
```

```python
x, y = coordenadas

x
```

```python
y
```

También se puede usar con listas:

```python
a, b, c = ['uno', 'dos', 'tres']

a
```

```python
b
```

## ¿Por qué existen tuplas y listas?

Aunque se podría programar solamente con listas, contar con ambos tipos de colecciones nos proporciona una gran versatilidad:

- **Desde un punto de vista semántico:** una lista representa una colección de datos que puede crecer y cambiar con el tiempo, mientras que una tupla representa una entidad con diversas partes. Por ejemplo:
	- **Listas:** listado de alumnos, productos en una cesta, órdenes procesadas.
	- **Tuplas:** una fecha `(año, mes, día)`, una coordenada en el espacio `(x, y, z)`, los parámetros de una distribución normal `(media, varianza)`.
- **Seguridad:** una tupla no se puede modificar, por lo que es idónea para proteger datos que no queremos modificar accidentalmente en el transcurso de un programa.
- **Hashable:** una tupla puede usarse como clave en un diccionario (es *hashable*), mientras que una lista no.
- **Rendimiento:** desde el punto de vista computacional, las tuplas son más sencillas, por lo que Python las procesa más rápidamente que las listas y consumen menos recursos.
