# Diccionarios

- Los diccionarios son estructuras de datos que almacenan pares clave:valor.
- Permiten acceder rápidamente a la información mediante la clave, de forma similar a un diccionario en la vida real.

```python
# Usando llaves
empleado = {'nombre': 'Ana', 'edad': 30, 'departamento': 'Contabilidad'}

# Usando dict()
producto = dict(nombre='Café', precio=2.5, stock=100)

# Diccionario vacío
vacio = { }
```

```python
vacio
```

## Actualizar con otro diccionario

El método `update()` permite añadir o modificar varios elementos a la vez utilizando otro diccionario.

```python
empleado.update({'cargo': 'Analista', 'edad': 32})
```

También con `|`

```python
empleado = empleado | {'cargo': 'Analista', 'edad': 32}
```

```python
empleado
```

## Iterar sobre un diccionario


```python
datos = {'a': 1, 'b': 2, 'c': 3}
```

Por clave:

```python
for clave in datos:
    print(clave)
```

Por valor:

```python
for valor in datos.values():
    print(valor)
```

Por clave y valor:

```python
for clave, valor in datos.items():
    print(clave, valor)
```

---
## Comprobar si una clave existe

Usando `in`.

```python
empleado = {'nombre': 'Ana', 'edad': 30 }

if 'nombre' in empleado:
    print('El empleado tiene nombre.')
```

## Métodos útiles


| Método                        | Descripción                                         |
|:------------------------------|:----------------------------------------------------|
| `keys()`                      | Devuelve todas las claves                           |
| `values()`                    | Devuelve todos los valores                          |
| `items()`                     | Devuelve pares (clave, valor)                       |
| `get(clave, defecto)`         | Accede sin lanzar `KeyError`                        |
| `setdefault(clave, defecto)`  | Devuelve el valor o lo inserta si no existe         |
| `pop(clave, defecto)`         | Elimina y devuelve el valor                         |
| `update(otro_dict)`           | Actualiza o añade elementos                         |
| `copy()`                      | Devuelve una copia superficial                      |
| `clear()`                     | Vacía el diccionario                                |
| `fromkeys(iterable, valor)`   | Crea un diccionario a partir de una lista de claves |

## Convertir dos listas en un diccionario


```python
claves =  ['nombre', 'edad', 'ciudad']
valores = [   'Ana',     30, 'Madrid']

persona = dict(zip(claves, valores))

persona
```

## Diccionarios por comprensión


```python
cuadrados = {x: x**2 for x in range(6)}

cuadrados
```

```python
pares = {x: x**2 for x in range(10) if x % 2 == 0}

pares
```

## Aplicaciones comunes

### Contar ocurrencias

```python
texto = 'banana'
conteo = {}

for letra in texto:
    conteo[letra] = conteo.get(letra, 0) + 1

conteo
```

### Agrupar elementos

```python
animales = ['perro', 'gato', 'pez', 'gaviota']
grupos = {}

for animal in animales:
    letra = animal[0]
    grupos.setdefault(letra, []).append(animal)

grupos
```
